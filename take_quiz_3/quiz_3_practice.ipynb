### 2. You will need to extract a subset of the data based on a logical condition. You will then group this subset, and aggregate one or more statistics. Then you will extract a subset of this data, based on where one of the aggregated statistics is greater than or equal to some value. Full credit will be given if you can do this in 1 line using chaining.

```python
df_results = pd.DataFrame({
    "raceId": [1, 1, 2, 2, 3, 3],
    "constructorId": [10, 20, 10, 20, 10, 20],
    "points": [10, 8, 6, 9, 12, 5]
})
```

```python
test_subset_group = (
    df_results.query("points >= 8")
    .groupby("constructorId")
    .agg(mean_points=("points", "mean"))
)
print(test_subset_group)
```

### 3. You will recode a numeric variable into a categorical variable (which takes string values). The category will be based on which interval the numeric variable belongs to.

```python
circuits["lat_category"] = np.where(
    circuits["lat"] < 0,
    "Southern Hemisphere",
    "Northern Hemisphere"
)
print(circuits[["circuit_name", "lat", "lat_category"]].head())
```

### 4. You will be asked to do the following: group a dataset by one of its columns/variables, aggregate one or more statistics for each group, and then sort the data by one of the statistics you aggregated over in descending order. Full credit will be given if you can do this in 1 line using chaining.

```python
df_sorted = (
    df_results.groupby("constructorId")["points"]
    .agg(avgpoints=("mean"))
    .sort_values(by="avgpoints", ascending=False)
)
print(df_sorted)
```
