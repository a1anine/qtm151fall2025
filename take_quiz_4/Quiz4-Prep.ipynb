### 3. Read in a .csv file to create a DataFrame which contains conversion rates for Euros to U.S. dollars and British Pounds to U.S. dollars over time. Compute non-dimensional percentage growth values for each and plot together on a single plot. (You can use matplotlib.pyplot or the .plot() method connected to a DataFrame).

```python
# Read CSV file
df = pd.read_csv("conversion_rates.csv")
df["date"] = pd.to_datetime(df["date"])
df.set_index("date", inplace=True)

# Calculate percentage growth (non-dimensional)
df["growth_euro"]  = df["euro_usd"].diff() / df["euro_usd"].shift() * 100
df["growth_pound"] = df["pound_usd"].diff() / df["pound_usd"].shift() * 100

# Plot both together
df[["growth_euro", "growth_pound"]].plot()

plt.xlabel("Date")
plt.ylabel("% Growth")
plt.title("Currency Conversion Growth (%)")
plt.legend(["Euro→USD", "Pound→USD"])
plt.show()
```

### 5. Continuing Problem 4, make a line plot of the conversion rate for this currency and highlight regions where the conversion rate is larger than a given value, using the function "plt.fill_between"

```python
# Condition for shading
condition = df["euro_usd"] > 1.10

# Line plot
plt.plot(df["date"], df["euro_usd"], color="blue", label="Euro→USD")

# Fill where condition True
plt.fill_between(
    x=df["date"],
    y1=df["euro_usd"],
    y2=df["euro_usd"].min(),
    where=condition,
    color="red",
    alpha=0.3,
    label="Above 1.10"
)

plt.xlabel("Date")
plt.ylabel("Euro→USD")
plt.title("Euro Conversion with Highlighted Regions")
plt.legend()
plt.show()
```
